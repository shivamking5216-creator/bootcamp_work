```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Sample data
data = {
    'Name': ['Amit', 'Priya', 'Rohan', 'Sneha', 'Karan'],
    'Study_Hour': [3, 2, 4, 5, 1],
    'Attendance': [85, 70, 90, 95, 60],
    'Previous_Marks': [60, 45, 75, 80, 40],
    'Final_Marks': [70, 50, 85, 90, 55]
}

df = pd.DataFrame(data)

# Extract only numeric columns for similarity
features = df[['Study_Hour', 'Attendance', 'Previous_Marks', 'Final_Marks']]

# Compute cosine similarity
cos_sim = cosine_similarity(features)

# Convert to DataFrame for readability
cos_sim_df = pd.DataFrame(cos_sim, index=df['Name'], columns=df['Name'])

print(cos_sim_df)
```
