```python
import pandas as pd
df = pd.read_csv('session4_dummy_students.csv')
df['Average'] = df[['Math', 'Science', 'English']].mean(axis=1)

df['Result'] = df['Average'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')

print("Class Average:", df['Average'].mean())

print("Highest Average:", df['Average'].max())
print("Lowest Average:", df['Average'].min())

top_avg = df['Average'].max()
top_students = df[df['Average'] == top_avg]
print(top_students[['Name', 'Average']])

low_attendance = df[df['Attendance'] < 75]
print(low_attendance[['Name', 'Attendance']])

few_assignments = df[df['Assignments'] < 6]
print(few_assignments[['Name', 'Assignments']])

df.to_csv('session4_student_report.csv', index=False)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url="https://raw.githubusercontent.com/plotly/datasets/master/gapminderDataFiveYear.csv"

df=pd.read_csv(url)


print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
year_min = df['year'].min()
year_max = df['year'].max()
print(f"Year range: {year_min} - {year_max}")

num_countries = df['country'].nunique()
print(f"Number of countries: {num_countries}")

# Number of continents
num_continents = df['continent'].nunique()
print(f"Number of continents: {num_continents}")
```

```python
# Group by continent and calculate average life expectancy
avg_life_exp = df.groupby('continent')['lifeExp'].mean()

# Plot bar chart
plt.figure(figsize=(8,6))
avg_life_exp.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Average Life Expectancy by Continent')
plt.xlabel('Continent')
plt.ylabel('Average Life Expectancy')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
summary=df.groupby('continent')[["pop","lifeExp", "gdpPercap"]].mean()
print(summary)
```

```python
# Filter data for India, USA, and China
countries = ["India", "United States", "China"]
df_subset = df[df['country'].isin(countries)]

# Plot line chart
plt.figure(figsize=(10,6))
for country in countries:
    country_data = df_subset[df_subset['country'] == country]
    plt.plot(country_data['year'], country_data['lifeExp'], marker='o', label=country)

plt.title('Life Expectancy Over Time (India, USA, China)')
plt.xlabel('Year')
plt.ylabel('Life Expectancy')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

# Scatter plot
plt.scatter(df['gdpPercap'], df['lifeExp'], alpha=0.6, c='skyblue', edgecolor='black')

plt.title('GDP per Capita vs Life Expectancy')
plt.xlabel('GDP per Capita')
plt.ylabel('Life Expectancy')
plt.grid(True)
plt.tight_layout()
plt.show()
```
